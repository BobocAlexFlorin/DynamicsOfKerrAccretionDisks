# Dynamics of Kerr Accretion Disks

# Phase 1: Relativistic Spacetime & Accretion Geometry

### Accretion onto rotating black holes is dictated by the Kerr metric. The spacetime geometry determines the location of the event horizon, the innermost stable circular orbit (ISCO), and the differential frame-dragging velocity field $\Omega(r, a)$ that drives orbital shear.

## 1.1 The Kerr Spacetime Geometry
In Boyer-Lindquist coordinates $(t, r, \theta, \phi)$ under natural units ($G = c = 1$), the line element of an uncharged, rotating black hole with mass $M$ and specific angular momentum $a = J/M \in [0, M)$ is:

$$ds^2 = -\left(1 - \frac{2Mr}{\rho^2}\right) dt^2 - \frac{4Mar\sin^2\theta}{\rho^2} dt\,d\phi + \frac{\rho^2}{\Delta} dr^2 + \rho^2 d\theta^2 + + \frac{\Sigma}{\rho^2} \sin^2\theta\, d\phi^2$$

### Metric Auxiliary Functions:

$$\rho^2(r, \theta) = r^2 + a^2 \cos^2\theta$$

$$\Delta(r) = r^2 - 2Mr + a^2$$

$$\Sigma(r, \theta) = \left(r^2 + a^2\right)^2 - a^2 \Delta(r) \sin^2\theta$$

## 1.2 Characteristic Astrophysical Radii

* **Outer Event Horizon ($r_+$):**  
  Point of no return; the outer coordinate singularity where $\Delta(r_+) = 0$:
  $$r_+ = M + \sqrt{M^2 - a^2}$$

* **Equatorial Ergosphere Boundary ($r_E$):**  
  The static limit ($g_{tt} = 0$ at $\theta = \pi/2$) where spacetime frame-dragging forces co-rotation:
  $$r_E = 2M$$

* **Innermost Stable Circular Orbit ($r_{\text{ISCO}}$):**  
  Inner edge of the thin accretion disk and boundary of stable circular prograde geodesics:
  $$r_{\text{ISCO}} = M \left( 3 + Z_2 - \sqrt{(3 - Z_1)(3 + Z_1 + 2Z_2)} \right)$$

  where:
  $$Z_1 = 1 + \left(1 - \frac{a^2}{M^2}\right)^{1/3} \left[\left(1 + \frac{a}{M}\right)^{1/3} + \left(1 - \frac{a}{M}\right)^{1/3}\right]$$
  $$Z_2 = \sqrt{3\frac{a^2}{M^2} + Z_1^2}$$

## 1.3 Relativistic Orbital Kinematics & Shear
For equatorial circular prograde geodesics ($\theta = \pi/2$), the relativistic Keplerian angular velocity observed at infinity is:
$$\Omega(r, a) = \frac{\sqrt{M}}{r^{3/2} + a\sqrt{M}}$$

The radial shear gradient $S(r, a)$ driving plasma turbulence and angular momentum transport is:
$$S(r, a) \equiv -r \frac{d\Omega}{dr} = \frac{3}{2} \frac{\sqrt{M} r^{3/2}}{(r^{3/2} + a\sqrt{M})^2}$$


# Phase 2: Magnetohydrodynamics & Instability Dynamics

### Pure Keplerian flow is Rayleigh-stable. Angular momentum is evacuated outward via the Balbus-Hawley Magnetorotational Instability (MRI), where weak poloidal magnetic fields couple neighboring plasma parcels.

## 2.1 The Relativistic MRI Dispersion Relation
In the local shearing box frame $(x, y, z)$, linear axisymmetric perturbations with vertical wavenumber $k_z$ embedded in a magnetic field $B_z$ obey:
$$\omega^4 - \omega^2 \left( \kappa_r^2 + 2\omega_A^2 \right) + \omega_A^2 \left( \omega_A^2 + \frac{d\Omega^2}{d\ln r} \right) = 0$$

### Characteristic Frequencies
Alfvén Wave Frequency:
$$\omega_A = k_z v_A = k_z \frac{B_z}{\sqrt{4\pi \rho_0}}$$

Relativistic Epicyclic Frequency (Kerr Metric):
$$\kappa_r^2(r, a) = \Omega^2(r, a) \left( 1 - \frac{6M}{r} + \frac{8a\sqrt{M}}{r^{3/2}} - \frac{3a^2}{r^2} \right)$$

### Dynamic Instability Criterion:
Dynamical growth ($\omega^2 < 0$) is triggered when:
$$\omega_A^2 < -\frac{d\Omega^2}{d\ln r} = -2\Omega r \frac{d\Omega}{dr}$$


# Phase 3: Second-Quantized Radial Lattice Model
### We discretize the continuous accretion disk outside the ISCO into $N$ concentric radial annuli. The localized plasma excitations are mapped to a tight-binding multi-mode lattice Hamiltonian.

## 3.1 Spatial Discretization Grid
$$r_j = r_{\text{ISCO}} + j \cdot \Delta r, \quad j \in \{0, 1, \dots, N-1\}$$

## 3.2 The Multi-Mode Plasma Hamiltonian
$$\hat{H} = \hat{H}_{\text{drift}} + \hat{H}_{\text{shear}} + \hat{H}_{\text{mag}}$$

$$\hat{H} = \sum_{j=0}^{N-1} \mu_j(a) \hat{a}_j^\dagger \hat{a}_j - \sum_{j=0}^{N-2} t_j(a) \left( \hat{a}_j^\dagger \hat{a}_{j+1} + \hat{a}_{j+1}^\dagger \hat{a}_j \right) + \sum_{j=0}^{N-2} U_j \hat{n}_j \hat{n}_{j+1}$$

## Coupling Parameters:

### On-Site Gravitational Drift ($\mu_j$):
$$\mu_j(a) = \Omega(r_j, a) = \frac{\sqrt{M}}{r_j^{3/2} + a\sqrt{M}}$$

### Shear-Induced Kinetic Hopping ($t_j$):
$$t_j(a) = t_0 \cdot S(r_j, a) = t_0 \frac{3\sqrt{M} r_j^{3/2}}{2(r_j^{3/2} + a\sqrt{M})^2}$$

### Maxwell Cross-Stress Interaction ($U_j$):
$$U_j \propto \frac{B_j^2}{8\pi}, \quad \hat{n}_j = \hat{a}_j^\dagger \hat{a}_j$$


# Phase 4: Jordan-Wigner Fermion-to-Qubit Isomorphism
### Quantum processors manipulate spin-1/2 Pauli operators. We map the creation and annihilation operators to Pauli strings while preserving the canonical anticommutation relations (CAR).

## 4.1 Isomorphism Transformations

$$\hat{a}_j^\dagger = \left( \bigotimes_{k=0}^{j-1} Z_k \right) \otimes \left( \frac{X_j - iY_j}{2} \right) \otimes \mathbb{I}^{\otimes (N-1-j)}$$

$$\hat{a}_j = \left( \bigotimes_{k=0}^{j-1} Z_k \right) \otimes \left( \frac{X_j + iY_j}{2} \right) \otimes \mathbb{I}^{\otimes (N-1-j)}$$

## Operator Mappings
### Quasiparticle Number Operator:
$$\hat{n}_j = \hat{a}_j^\dagger \hat{a}_j = \frac{\mathbb{I} - Z_j}{2}$$

### Kinetic Hopping Operator:
$$\hat{a}_j^\dagger \hat{a}_{j+1} + \hat{a}_{j+1}^\dagger \hat{a}_j = -\frac{1}{2} \left( X_j X_{j+1} + Y_j Y_{j+1} \right)$$

### Non-Linear Maxwell Density-Density Interaction:
$$\hat{n}_j \hat{n}_{j+1} = \frac{1}{4} \left( \mathbb{I} - Z_j - Z_{j+1} + Z_j Z_{j+1} \right)$$


## 4.2 The Full Qubit Pauli Hamiltonian
$$\hat{H}_q = C_0 \mathbb{I} + \sum_{j=0}^{N-1} h_j Z_j + \sum_{j=0}^{N-2} \frac{t_j(a)}{2} \left( X_j X_{j+1} + Y_j Y_{j+1} \right) + \sum_{j=0}^{N-2} \frac{U_j}{4} Z_j Z_{j+1}$$

### Effective Shift and Local Field Coefficients
$$C_0 = \sum_{j=0}^{N-1} \frac{\mu_j(a)}{2} + \sum_{j=0}^{N-2} \frac{U_j}{4}$$

$$h_j = -\frac{\mu_j(a)}{2} - \frac{U_j + U_{j-1}}{4}, \quad \text{with } U_{-1} \equiv 0 \text{ and } U_{N-1} \equiv 0$$


# Phase 5: Variational Quantum Eigensolver (VQE)
### We prepare the stable magnetorotational ground state $\vert\psi_0\rangle$ of the accretion disk on a quantum register by optimizing a parameterized quantum ansatz $U(\vec{\theta})$.

## 5.1 The Rayleigh-Ritz Variational Formulation
$$E(\vec{\theta}) = \langle \psi(\vec{\theta}) \vert \hat{H}_q \vert \psi(\vec{\theta}) \rangle = \sum_k c_k \langle \psi(\vec{\theta}) \vert \hat{P}_k \vert \psi(\vec{\theta}) \rangle \ge E_0$$

### Particle-Conserving Givens Excitations ($SO(4)$ Rotation):
To conserve total excitation count $\sum_j \hat{n}_j = N_{\text{excitations}}$, we use parameter blocks generated by Givens transformations:
$$G(\theta) = \exp\left( -i \frac{\theta}{2} (X_j Y_{j+1} - Y_j X_{j+1}) \right)$$

## 5.2 Analytical Gradient Evaluation: The Parameter-Shift Rule
When evaluating gradients over single-Pauli generator primitives or strictly within the single-excitation subspace:
$$\frac{\partial E(\vec{\theta})}{\partial \theta_k} = \frac{E\left(\vec{\theta} + \frac{\pi}{2}\hat{e}_k\right) - E\left(\vec{\theta} - \frac{\pi}{2}\hat{e}_k\right)}{2}$$


# Phase 6: Unitary Real-Time Dynamics & Quantum Simulation
### We simulate non-equilibrium turbulence, magnetic reconnection bursts, or wave packet propagation by evolving the quantum state forward in time under $\hat{U}(t) = e^{-i \hat{H}_q t}$.

## 6.1 Suzuki-Trotter Product Decomposition
Because terms in $\hat{H}_q = \sum_k c_k \hat{P}_k$ generally do not commute ($[\hat{P}_i, \hat{P}_j] \neq 0$), the time-evolution operator is split into $n$ discrete Trotter time steps ($\Delta t = t/n$):

### 1st-Order Lie-Trotter Formula:
$$U_1(t) = \left( \prod_k e^{-i c_k \hat{P}_k \Delta t} \right)^n, \quad \text{Error: } \mathcal{O}\left(\frac{t^2}{n}\right)$$

### 2nd-Order Symmetric Suzuki-Trotter Formula:
$$U_2(t) = \left( \prod_{k=1}^M e^{-i c_k \hat{P}_k \frac{\Delta t}{2}} \prod_{k=M}^1 e^{-i c_k \hat{P}_k \frac{\Delta t}{2}} \right)^n, \quad \text{Error: } \mathcal{O}\left(\frac{t^3}{n^2}\right)$$


# Phase 7: Astrophysical Observables & Visual Signatures
### Quantum measurement outputs from the evolved register map directly to physical accretion observables: wave packet dispersion, survival fidelity, and the broad relativistic iron $\text{Fe K}\alpha$ emission line.

## 7.1 Quantum Fidelity & Radial Density
### Quantum State Survival Fidelity:
$$F(t) = \left\vert \langle \psi(0) \vert \psi(t) \rangle \right\vert^2$$

### Time-Resolved Radial Plasma Density:
$$\langle n_j(t) \rangle = \frac{1 - \langle Z_j(t) \rangle}{2}$$

## 7.2 Relativistic Iron $\text{Fe K}\alpha$ Emission Line Profile
The photon energy shift factor $g$ for emission originating at radius $r_j$ and azimuth $\phi$, observed at inclination angle $i$, is:

$$g(r_j, \phi) = \frac{E_{\text{obs}}}{E_{\text{em}}} = \frac{\sqrt{1 - \frac{3M}{r_j} + \frac{2a\sqrt{M}}{r_j^{3/2}}}}{1 + \Omega(r_j, a) \sqrt{r_j^2 + a^2} \sin i \sin \phi}$$

The aggregate observed relativistic spectral line flux profile is:
$$F(E_{\text{obs}}) = \sum_{j=0}^{N-1} \int_0^{2\pi} \langle n_j(t) \rangle \cdot g(r_j, \phi)^3 \, \delta\left( E_{\text{obs}} - g(r_j, \phi) E_0 \right) r_j \Delta r \, d\phi$$

```python
# Imports

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

```python
# Quantum Circuit

class HardwareQuantumCircuit:
    # Statevector simulator using 1-qubit & 2-qubit logic gates
    def __init__(self, num_qubits):
        self.num_qubits = num_qubits
        self.dim = 1 << num_qubits
        self.state = np.zeros(self.dim, dtype=complex)
        self.state[0] = 1.0 # |0000...0>

    def reset_to_state(self, state_vec):
        self.state = state_vec.copy()

    def x(self, q):
        # Pauli X-Gate
        shape = [2] * self.num_qubits
        tensor = self.state.reshape(shape)
        tensor = np.swapaxes(tensor, 0, q)
        tensor[:] = tensor[::-1]
        tensor = np.swapaxes(tensor, 0, q)
        self.state = tensor.flatten()

    def h(self, q):
        # Hadamard Gate
        H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2.0)
        self._apply_1q_gate(H, q)

    def s(self, q, dagger=False):
        # Phase gate S/S^dagger
        phase = -1j if dagger else 1j
        S = np.array([[1, 0], [0, phase]], dtype=complex)
        self._apply_1q_gate(S, q)

    def rz(self, theta, q):
        # Rz(theta) = e^(-i * theta/2 * Z)
        R = np.array([[np.exp(-1j * theta / 2.0), 0], [0, np.exp(1j * theta / 2.0)]], dtype=complex)
        self._apply_1q_gate(R, q)

    def ry(self, theta, q):
        # Ry(theta) = e^(-i * theta/2 * Y)
        R = np.array([[np.cos(theta / 2.0), -np.sin(theta / 2.0)], [np.sin(theta / 2.0), np.cos(theta / 2.0)]], dtype=complex)
        self._apply_1q_gate(R, q)

    def cnot(self, control, target):
        #CNOT gate
        shape = [2] * self.num_qubits
        tensor = self.state.reshape(shape)
        tensor = np.swapaxes(tensor, 0, control)
        tensor = np.swapaxes(tensor, 1, target)

        # Invert target when control = 1
        tensor[1] = tensor[1, ::-1]

        tensor = np.swapaxes(tensor, 1, target)
        tensor = np.swapaxes(tensor, 0, control)
        self.state = tensor.flatten()


    def _apply_1q_gate(self, gate, q):
        shape = [2] * self.num_qubits
        tensor = self.state.reshape(shape)
        tensor = np.swapaxes(tensor, 0, q)
        tensor = np.tensordot(gate, tensor, axes=(1, 0))
        tensor = np.swapaxes(tensor, 0, q)
        self.state = tensor.flatten()

    # Transpilation Methods

    def givens_rotation(self, theta, q0, q1):
        # Transpiles particle conservating SO(4) Given rot -> CNOT & Ry Gates
        self.cnot(q1, q0)
        self.ry(-theta, q1)
        self.cnot(q0, q1)
        self.ry(theta, q1)
        self.cnot(q0, q1)
        self.cnot(q1, q0)

    def apply_xx_yy_hop(self, alpha, q0, q1):
        # Transpiles hopping e^(-i * alpha * (XX + YY) -> native gates)
        # 1. exp(-i * alpha * XX)
            self.h(q0)
            self.h(q1)
            self.cnot(q0, q1)
            self.rz(2.0 * alpha, q1)
            self.cnot(q0, q1)
            self.h(q0)
            self.h(q1)
            
            # 2. exp(-i * alpha * YY)
            self.s(q0, dagger=True)
            self.s(q1, dagger=True)
            self.h(q0)
            self.h(q1)
            self.cnot(q0, q1)
            self.rz(2.0 * alpha, q1)
            self.cnot(q0, q1)
            self.h(q0)
            self.h(q1)
            self.s(q0)
            self.s(q1)

    def apply_zz_coupling(self, alpha, q0, q1):
            #Transpiles Maxwell interaction e^(-i * alpha * ZZ)
            self.cnot(q0, q1)
            self.rz(2.0 * alpha, q1)
            self.cnot(q0, q1)
```

```python
# 2D Kerr Accretion Disk Lattice

class Kerr2DAccretionLattice:
    def __init__(self, M=1.0, a=0.95, Nr=3, Nphi=2, dr=0.8, t0_r=1.0, t0_phi=0.8, U_mag=0.2):
        self.M, self.a = float(M), float(a)
        self.Nr, self.Nphi = int(Nr), int(Nphi)
        self.num_qubits = self.Nr * self.Nphi
        self.dr = float(dr)
        self.t0_r, self.t0_phi = float(t0_r), float(t0_phi)
        self.U_mag = float(U_mag)

        # ISCO and Horizon
        self.r_plus = self.M + np.sqrt(self.M**2 - self.a**2)
        z1 = 1.0 + (1.0 - self.a**2 / self.M**2)**(1/3) * (
            (1.0 + self.a / self.M)**(1/3) + (1.0 - self.a / self.M)**(1/3)
        )
        z2 = np.sqrt(3.0 * (self.a**2 / self.M**2) + z1**2)
        self.r_isco = self.M * (3.0 + z2 - np.sqrt((3.0 - z1) * (3.0 + z1 + 2.0 * z2)))

        self.radii = np.array([self.r_isco + j * self.dr for j in range(self.Nr)])
        self.dphi = 2.0 * np.pi / self.Nphi
        self.phi_grid = np.array([k * self.dphi for k in range(self.Nphi)])

        # Relativistic Kinematics
        self.Omega = np.sqrt(self.M) / (self.radii**1.5 + self.a * np.sqrt(self.M))
        self.shear = 1.5 * np.sqrt(self.M) * (self.radii**1.5) / (
            (self.radii**1.5 + self.a * np.sqrt(self.M))**2
        )

        # Couplings
        self.t_r = self.t0_r * self.shear[:-1]
        self.t_phi = self.t0_phi * (self.radii * self.Omega) / self.dphi

    def q_idx(self, j, k):
        #Maps radial_ring and azimuthal_sector to a linear qubit index
        return j * self.Nphi + (k % self.Nphi)
```

```python
# 2D Native-Gate VQE & Trotterization

class NativeGateKerrSolver:
    def __init__(self, lattice_2d):
        self.lat = lattice_2d
        self.N = lattice_2d.num_qubits

    def prepare_vqe_ansatz(self, theta_vec):
        #Constructs particle-conserving trial state using 2D native Givens circuits
        qc = HardwareQuantumCircuit(self.N)
        # Initialize 1 excitation at the innermost ring (j=0, k=0)
        qc.x(self.lat.q_idx(0, 0))

        param_idx = 0
        # Layer 1: Radial Givens rotations (j <-> j+1)
        for j in range(self.lat.Nr - 1):
            for k in range(self.lat.Nphi):
                q0 = self.lat.q_idx(j, k)
                q1 = self.lat.q_idx(j + 1, k)
                qc.givens_rotation(theta_vec[param_idx], q0, q1)
                param_idx += 1

        # Layer 2: Azimuthal Givens rotations (k <-> k+1 periodic)
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q0 = self.lat.q_idx(j, k)
                q1 = self.lat.q_idx(j, k + 1)
                if q0 < q1 or self.lat.Nphi > 2:
                    qc.givens_rotation(theta_vec[param_idx], q0, q1)
                    param_idx += 1

        return qc.state

    def measure_expectation_energy(self, state_vec):
        #Evaluates total energy <psi|H|psi> across 2D radial and azimuthal couplings
        energy = 0.0
        # On-site drift & Maxwell shift
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q = self.lat.q_idx(j, k)
                prob_1 = np.sum(np.abs(state_vec[[(idx & (1 << (self.N - 1 - q))) != 0 for idx in range(1 << self.N)]])**2)
                energy += self.lat.Omega[j] * prob_1

        # Hopping energies
        # Radial:
        for j in range(self.lat.Nr - 1):
            for k in range(self.lat.Nphi):
                q0, q1 = self.lat.q_idx(j, k), self.lat.q_idx(j + 1, k)
                energy -= 2.0 * self.lat.t_r[j] * self._get_hopping_overlap(state_vec, q0, q1)

        # Azimuthal:
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q0, q1 = self.lat.q_idx(j, k), self.lat.q_idx(j, k + 1)
                if q0 != q1:
                    energy -= 2.0 * self.lat.t_phi[j] * self._get_hopping_overlap(state_vec, q0, q1)
        return energy

    def _get_hopping_overlap(self, state, q0, q1):
        """Computes Re(<psi| a_q0^dagger a_q1 |psi>)."""
        mask0 = 1 << (self.N - 1 - q0)
        mask1 = 1 << (self.N - 1 - q1)
        val = 0.0
        for i in range(1 << self.N):
            if (i & mask0) and not (i & mask1):
                j = (i ^ mask0) | mask1
                val += np.real(np.conj(state[i]) * state[j])
        return val

    def solve_ground_state(self):
        # Determine number of variational parameters
        num_params = (self.lat.Nr - 1) * self.lat.Nphi + self.lat.Nr * (self.lat.Nphi if self.lat.Nphi > 2 else 1)
        init_theta = np.random.uniform(-0.1, 0.1, num_params)

        res = minimize(lambda th: self.measure_expectation_energy(self.prepare_vqe_ansatz(th)), 
                       init_theta, method='L-BFGS-B')
        psi_0 = self.prepare_vqe_ansatz(res.x)
        return psi_0, res.fun

    def trotter_step_circuit(self, qc, dt):
        #Executes one second-order symmetric Trotter step using native logic gates
        half_dt = dt / 2.0

        # 1. On-site Rz gates
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q = self.lat.q_idx(j, k)
                qc.rz(self.lat.Omega[j] * half_dt, q)

        # 2. Radial Hopping XX + YY
        for j in range(self.lat.Nr - 1):
            for k in range(self.lat.Nphi):
                q0 = self.lat.q_idx(j, k)
                q1 = self.lat.q_idx(j + 1, k)
                qc.apply_xx_yy_hop(-self.lat.t_r[j] * dt / 2.0, q0, q1)

        # 3. Azimuthal Hopping XX + YY
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q0 = self.lat.q_idx(j, k)
                q1 = self.lat.q_idx(j, k + 1)
                if q0 != q1:
                    qc.apply_xx_yy_hop(-self.lat.t_phi[j] * dt / 2.0, q0, q1)

        # 4. On-site Rz gates (Symmetric completion)
        for j in range(self.lat.Nr):
            for k in range(self.lat.Nphi):
                q = self.lat.q_idx(j, k)
                qc.rz(self.lat.Omega[j] * half_dt, q)
```

```python
# Simulation & 2D Observational Spectra

if __name__ == "__main__":
    np.random.seed(42)
    print("=" * 70)
    print("2D HARDWARE-COMPILED KERR ACCRETION DISK QUANTUM SIMULATION")
    print("=" * 70)

    # 1. Initialize 2D Grid: Nr=3 rings, Nphi=2 sectors = 6 Qubits
    lat_2d = Kerr2DAccretionLattice(M=1.0, a=0.95, Nr=3, Nphi=2, dr=0.8, t0_r=0.9, t0_phi=1.1)
    solver = NativeGateKerrSolver(lat_2d)

    print(f"[*] Total Qubits:      {lat_2d.num_qubits} (Grid: {lat_2d.Nr} rings x {lat_2d.Nphi} azimuthal sectors)")
    print(f"[*] ISCO Radius:       {lat_2d.r_isco:.4f} M")
    print(f"[*] Rings (r):         {np.round(lat_2d.radii, 3)} M")

    # 2. Run Native VQE
    psi_0, E_0 = solver.solve_ground_state()
    print(f"[*] Native VQE Ground Energy: {E_0:.6f}")

    # 3. Inject Relativistic Magnetic Flare Perturbation
    qc_dyn = HardwareQuantumCircuit(lat_2d.num_qubits)
    qc_dyn.reset_to_state(psi_0)
    # Phase kick on innermost azimuthal parcel (j=0, k=0)
    qc_dyn.rz(np.pi, lat_2d.q_idx(0, 0))
    psi_perturbed = qc_dyn.state.copy()

    # 4. Run Trotterized Unitary Time Evolution
    time_steps = 80
    t_max = 12.0
    dt = t_max / time_steps
    time_axis = np.linspace(0, t_max, time_steps)

    densities_2d = np.zeros((lat_2d.Nr, lat_2d.Nphi, time_steps))
    fidelities = []

    for step in range(time_steps):
        # Compute survival fidelity
        fid = np.abs(np.vdot(psi_perturbed, qc_dyn.state))**2
        fidelities.append(fid)

        # Measure 2D densities <n_{j,k}>
        for j in range(lat_2d.Nr):
            for k in range(lat_2d.Nphi):
                q = lat_2d.q_idx(j, k)
                mask = 1 << (lat_2d.num_qubits - 1 - q)
                prob = np.sum(np.abs(qc_dyn.state[[(idx & mask) != 0 for idx in range(1 << lat_2d.num_qubits)]])**2)
                densities_2d[j, k, step] = prob

        # Step forward using compiled hardware gates
        solver.trotter_step_circuit(qc_dyn, dt)
        E_bins = np.linspace(3.0, 8.5, 180)
    flux_2d = np.zeros(len(E_bins))
    sin_i = np.sin(np.radians(50.0))

    # Integrate emission over the evolved 2D disk
    evolved_dens = densities_2d[:, :, 40]  # Midpoint of time evolution
    for j, r in enumerate(lat_2d.radii):
        u_t = 1.0 / np.sqrt(max(1.0 - 3.0 * lat_2d.M / r + 2.0 * lat_2d.a * np.sqrt(lat_2d.M) / (r**1.5), 1e-6))
        for k, phi in enumerate(lat_2d.phi_grid):
            weight = evolved_dens[j, k] * r * lat_2d.dr * lat_2d.dphi
            g = 1.0 / (u_t * (1.0 + lat_2d.Omega[j] * np.sqrt(r**2 + lat_2d.a**2) * sin_i * np.sin(phi)))
            E_obs = g * 6.4
            idx = int((E_obs - E_bins[0]) / (E_bins[1] - E_bins[0]))
            if 0 <= idx < len(E_bins):
                flux_2d[idx] += weight * (g**3)

    # Plot Diagnostics
    plt.style.use('dark_background')
    fig, axes = plt.subplots(1, 3, figsize=(18, 5), dpi=120)

    # 1. 2D Accretion Density Heatmap (Summed over azimuth)
    radial_density_t = np.sum(densities_2d, axis=1)
    im = axes[0].imshow(radial_density_t, aspect='auto', origin='lower',
                        extent=[time_axis[0], time_axis[-1], lat_2d.radii[0], lat_2d.radii[-1]], cmap='magma')
    axes[0].set_title(f"2D Shear & Orbital Dispersion ($a={lat_2d.a}$)", fontsize=11)
    axes[0].set_xlabel("Coordinate Time $t$ ($M$)")
    axes[0].set_ylabel("Accretion Radius $r$ ($M$)")
    plt.colorbar(im, ax=axes[0], label=r"Total Radial Density $\sum_k \langle n_{j,k}(t) \rangle$")

    # 2. Fidelity Decoherence
    axes[1].plot(time_axis, fidelities, color='#00e5ff', lw=2.2)
    axes[1].set_title("Flare Survival Fidelity $F(t)$", fontsize=11)
    axes[1].set_xlabel("Coordinate Time $t$ ($M$)")
    axes[1].set_ylabel(r"$|\langle \psi(0) | \psi(t) \rangle|^2$")
    axes[1].grid(True, linestyle=':', alpha=0.4)

    # 3. Relativistic 2D Fe K-alpha Spectral Profile
    axes[2].plot(E_bins, flux_2d, color='#ffab00', lw=2.2, label=r"2D Disk Emission ($i=50^\circ$)")
    axes[2].axvline(6.4, color='white', linestyle=':', alpha=0.5, label="Rest Frame (6.4 keV)")
    axes[2].set_title(r"Relativistic $\mathrm{Fe\ K}\alpha$ Profile from 2D Register", fontsize=11)
    axes[2].set_xlabel(r"Observed Photon Energy $E_{\mathrm{obs}}$ (keV)")
    axes[2].set_ylabel("Normalized Flux")
    axes[2].legend(frameon=True, facecolor='#1e1e1e', edgecolor='none')
    axes[2].grid(True, linestyle=':', alpha=0.4)

    plt.tight_layout()
    plt.show()
    print("\n2D Quantum pipeline and native circuit transpilation executed successfully.")
```
